### Problem 10: Percentage Function

Write a function `to_percentage(correct, total)` that returns the percentage score (`correct / total * 100`). Test it with at least two different pairs of numbers.

````python
# TODO: write your code here
def to_percentage(correct, total):
  if total == 0:
    return 0.0
  return (correct / total) * 100


# Test cases
print(f"Score 1: {to_percentage(45, 50)}%")  # Test case 1
print(f"Score 2: {to_percentage(17, 20)}%")  # Test case 2
print(f"Score 3: {to_percentage(7, 10)}%")  # Additional test case
````

### Problem 19: Mini Vocabulary Quiz (one question)

Create two variables: `question_word` (e.g. `"apple"`) and `correct_meaning` (e.g. `"사과"`). Ask the user to type the meaning using `input()`. Write a function `check_answer(user_answer, correct_answer)` that returns `True` or `False` (ignore uppercase/lowercase and extra spaces using `.strip().lower()`). Print `"Correct!"` if the function returns `True`, otherwise print `"Try again!"`.

💡 **Hint — combining a function with `if`:**
```python
def is_even(number):
    return number % 2 == 0

n = 4

    print("Even!")
else:
    print("Odd!")
```


````python
# TODO: write your code here
# Create the variables
question_word = "apple"
correct_meaning = "사과"

# Write the checking function
def check_answer(user_answer, correct_answer):
    # Normalize inputs by removing extra spaces and ignoring case
    return user_answer.strip().lower() == correct_answer.strip().lower()

# Ask the user to type the meaning
user_input = input(f"What is the meaning of '{question_word}'? ")

# Check the answer and print the result
if check_answer(user_input, correct_meaning):
    print("Correct!")
else:
    print("Try again!")
````

````python
# TODO: write your code here
import random

# 1. Create word and meaning lists
word_list = ["apple", "book", "cat", "dog", "house"]
meaning_list = ["사과", "책", "고양이", "개", "집"]


# 2. Write the check_answer function
def check_answer(user_answer, correct_answer):
  # Strip whitespace for safety and compare
  return user_answer.strip() == correct_answer.strip()


# Initialize the score variable
score = 0

# 3. Run the quiz loop 3 times
for _ in range(3):
  # Pick a random index between 0 and 4
  index = random.randint(0, len(word_list) - 1)

  word = word_list[index]
  correct_meaning = meaning_list[index]

  # Ask the user for the meaning
  user_answer = input(f"What is the meaning of '{word}'? ")

  # 4. Check the answer and update the score
  if check_answer(user_answer, correct_meaning):
    print("Correct! 🎉")
    score += 1
  else:
    print(f"Incorrect. The correct answer is '{correct_meaning}'.")

# 5. Print the final score
print(f"Final score: {score}/3")
````
